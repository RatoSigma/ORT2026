```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
df = pd.read_csv('houses_to_rent.csv')
display(df.head())
df.info()
display(df.isnull().sum())
```

```python
colunas_valores = ['hoa', 'rent amount', 'property tax', 'fire insurance', 'total']

for coluna in colunas_valores:
    df[coluna] = (
        df[coluna]
        .astype(str)
        .str.replace('R$', '', regex=False)
        .str.replace(',', '', regex=False)
        .str.strip()
        .replace({'Sem info': '0', 'Incluso': '0'})
        .astype(float)
    )

display(df[colunas_valores].head())
```

```python
df['floor'] = df['floor'].replace('-', 0).astype(int)
df['animal'] = df['animal'].replace({'acept': 1, 'not acept': 0})
df['furniture'] = df['furniture'].replace({'furnished': 1, 'not furnished': 0})

if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

display(df.head())
df.info()
```

```python
X = df.drop(columns=['total'])
y = df['total']

modelo = LinearRegression()
modelo.fit(X, y)

delta = pd.Series(modelo.coef_, index=X.columns)

print('Intercepto:', modelo.intercept_)
print(delta)
```

```python
entrada = {
    'city': 1,
    'area': 240,
    'rooms': 3,
    'bathroom': 3,
    'parking spaces': 4,
    'floor': 0,
    'animal': 1,
    'furniture': 1,
    'hoa': 0.0,
    'rent amount': 8000.0,
    'property tax': 1000.0,
    'fire insurance': 121.0
}

imovel = pd.DataFrame([entrada])
previsao = modelo.predict(imovel)

print('Valor total previsto do aluguel: R$', round(previsao[0], 2))
```
