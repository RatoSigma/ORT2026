```python
# Parte 2
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style='darkgrid')
```

```python
# Carregando a base de carros
df_mpg = sns.load_dataset('mpg')
display(df_mpg.head())
```

```python
plt.figure(figsize=(8, 4))

sns.scatterplot(data=df_mpg, x='horsepower', y='mpg', hue='origin', palette='Set2')

plt.title("Gráfico de Disperção entre Potência e MPG")
plt.xlabel("Potência")
plt.ylabel("Milhas por Galão (MPG)")
plt.show()
```

```python
plt.figure(figsize=(8, 4))

sns.lineplot(data=df_mpg, x='mpg', y='model_year')

plt.title("Gráfico de Linha entre MPG e Ano")
plt.xlabel("MPG (Milhas por Galão)")
plt.ylabel("Modelo de Ano")
plt.show()
```

```python
plt.figure(figsize=(8, 4))

sns.boxplot(data=df_mpg, x='acceleration', y='origin', hue='origin', palette='Set2')

plt.title("Gráfico Comparativo de Aceleração por Origem")
plt.xlabel("Origem")
plt.ylabel("Aceleração")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

colunas_numericas = df_mpg[['mpg', 'cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model_year']]

matriz_corr = colunas_numericas.corr()

sns.heatmap(matriz_corr, annot=True, cmap='coolwarm', fmt='.2f',
linewidths=0.5)
plt.title('Matriz de Correlação de Variáveis Numéricas')
plt.show()
```
