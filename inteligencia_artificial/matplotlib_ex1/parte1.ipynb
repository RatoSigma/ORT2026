```python
# Parte 1

import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style='darkgrid')
```

```python
# Carregando a base do Titanic
df_titanic = sns.load_dataset('titanic')
display(df_titanic .head())
```

```python
sns.histplot(data=df_titanic, x='fare', kde=True, bins=30, color='teal')

plt.title("Gráfico de Distribuição de Tarifas")
plt.xlabel("Quantidade")
plt.ylabel("Tarifa")
```

```python
sns.boxplot(data=df_titanic, x='sex', y='age')

plt.title("Gráfico Comparativo de Idade por Gênero")
plt.xlabel("Sexo do Passageiro")
plt.ylabel("Idade")
```

```python
sns.regplot(data=df_titanic, x='age', y='fare')

plt.title("Gráfico de Tendência entre Idade e Tarifa")
plt.xlabel("Idade")
plt.ylabel("Tarifa")
```
